# Question 3 (12 Points)

In this question we consider the following point operation $M: [0,1] \to [0,1]$:

```python
def M(v, k1, k2):
    """Apply the point operation to a single gray level v."""
    if not 0 <= k1 < k2 <= 1:
        raise ValueError("Expected 0 <= k1 < k2 <= 1.")
    if v <= k1:
        return 0
    if v >= k2:
        return 1
    return (v - k1) / (k2 - k1)
```

Let $f$ be a grayscale image with values in $[0,1]$, and let $g(x) = M(f(x))$ be the image obtained by applying $M$ pixelwise.

## 1. Invariance

Does there exist an image $f$ with at least three distinct gray levels that is invariant under $M$ (i.e., $M(f) = f$ pointwise)? If yes, give an example and prove it. If not, prove that no such image exists.

### <span style="color: green;">Answer</span>

<div style="color: green;">

Yes. When $k_1=0$ and $k_2=1$, $M$ is the identity operation:
- $v=0$: $M(0)=0$.
- $0<v<1$: $M(v)=\frac{v-0}{1-0}=v$.
- $v=1$: $M(1)=1$.

For other values of $k_1$ and $k_2$ that satisfy $0 < k_1 < k_2 < 1$, we will choose an appropriate $v$ so that $M(v)=v$:
- $\frac{v - k_1}{k_2 - k_1} = v \implies v - k_1 = v(k_2 - k_1) \implies v(1 + k_1 - k_2) = k_1 \implies v = \frac{k_1}{1 + k_1 - k_2}$

Thus, $M(f)=f$ for any image, including one with at least three distinct gray levels (0,1, and at least one value $v$ that satisfies: $k_1<v<k_2$)
</div>

## 2. Mean intensity

Let $\mu(f)$ denote the mean gray level of $f$. How does $\mu(g)$ compare to $\mu(f)$? Is it always larger, always smaller, or dependent on $f$ (and on $k_1,k_2$)? Explain.

### <span style="color: green;">Answer</span>



<div style="color: green;">

The mean depends on $f$, $k_1$, and $k_2$. For example, with $k_1=0.4$ and $k_2=0.7$:
- Decrease: if $f=0.3$ then $\mu(f)=0.3>\mu(g)=0$.
- Increase: if $f=0.75$ then $\mu(f)=0.75<\mu(g)=1$.

The point operation stretches pixel values linearly from 0 to 1, and clips values below $k_1$ to 0 (greenucing the mean), and values above $k_2$ are mapped to 1 (increasing the mean). Whether the mean of the mapped image will increase or decrease depends on where is most of the pixel mass relative to $k_1$ and $k_2$.

</div>


## 3. Contrast & iteration

Assume for the sake of this section that $f$ contains values only in $[k_1,k_2]$, such that $g \ne f$.

### (a) Contrast

Compare $\operatorname{var}(g)$ and $\operatorname{var}(f)$ (contrast). Is $\operatorname{var}(g)$ always larger, always smaller, or dependent on $f$? Justify.

#### <span style="color: green;">Answer</span>

<div style="color: green;">

$\operatorname{var}(g)$ is always larger or equal to $\operatorname{var}(f)$

Since all values of $f$ lie in $[k_1,k_2]$,

$$
g=\frac{f-k_1}{k_2-k_1}
\qquad
\mu(g)=\frac{\mu(f)-k_1}{k_2-k_1}
$$

Subtracting the mean gives

$$
g-\mu(g)
=\frac{f-k_1}{k_2-k_1}
-\frac{\mu(f)-k_1}{k_2-k_1}
=\frac{f-\mu(f)}{k_2-k_1}
$$

Therefore,

$$
\begin{aligned}
\operatorname{var}(g)
&=\mathbb{E}\!\left[(g-\mu(g))^2\right]\\
&=\mathbb{E}\!\left[\left(\frac{f-\mu(f)}{k_2-k_1}\right)^2\right]\\
&=\frac{1}{(k_2-k_1)^2}\,
  \mathbb{E}\!\left[(f-\mu(f))^2\right]\\
&=\frac{\operatorname{var}(f)}{(k_2-k_1)^2}
\end{aligned}
$$

Since $0<k_2-k_1\le1$,

$$
\frac{1}{(k_2-k_1)^2}\ge1
\quad\Longrightarrow\quad
\operatorname{var}(g)\ge\operatorname{var}(f)
$$

Example: For $k_1=0.2$ and $k_2=0.7$:

- **Constant image:** If $f=0.3$, then $g=0.2$. Both variances are zero.
- **Nonconstant image:** The variance strictly increases.

 For equally frequent levels $f=\{0.3,0.45,0.6\}$, we get $g=\{0.2,0.5,0.8\}$, so $\operatorname{var}(g)=4\operatorname{var}(f)$.


</div>


### (b) Iteration

Consider repeated application $f^{(t+1)} = M(f^{(t)})$ with $f^{(0)} = f$. Describe the limiting image as $t \to \infty$.

#### <span style="color: green;">Answer</span>

<div style="color: green;">
At infinity, we will get either a binary image or an image with 3 grey levels.

The threshold is the fixed point satisfying:

$$
M(\theta)=\theta
\quad\Longrightarrow\quad
\frac{\theta-k_1}{k_2-k_1}=\theta
\quad\Longrightarrow\quad
\theta=\frac{k_1}{1-k_2+k_1}
$$

 Repeated application converges to:

$$
M_\infty(v)=
\begin{cases}
0, & v<\theta\\
\theta, & v=\theta\\
1, & v>\theta
\end{cases}
$$

Thus, pixels converge to $0$ or $1$, except those exactly at $\theta$, which remain unchanged.

</div>


## 4. Negative after application

Does there exist an image $h$ such that $M(h) = 1-h$? If yes—construct and prove. If not—prove impossibility.

### <span style="color: green;">Answer</span>

<div style="color: green;">

Yes. We select a constant image $h\equiv v$, where

$$
M(v)=1-v
\quad\Longrightarrow\quad
\frac{v-k_1}{k_2-k_1}=1-v
\quad\Longrightarrow\quad
v=\frac{k_2}{1+k_2-k_1}
$$

This value lies in $(k_1,k_2)$, so the middle branch applies and $M(h)=1-h$ pointwise.

**Example:** For $k_1=0.2$, $k_2=0.8$, and $h\equiv0.5$,

$$
M(h)=\frac{0.5-0.2}{0.8-0.2}=0.5=1-h
$$

</div>

# Question 4 (8 Points)

Let $I_1$ be a grayscale image. First apply histogram equalization to obtain $I_2$. Then apply a point operation on $I_2$ to obtain $I_3$.


## 1. Overall point operation

Prove that the overall mapping from $I_1$ to $I_3$ is itself a point operation. That is, there exists a function $M$ on gray levels such that $I_3(x)=M(I_1(x))$ for all pixels $x$.

### <span style="color: green;">Answer</span>


<div style="color: green;">

For the given image $I_1$, histogram equalization defines a gray level mapping $M_1$ - Because histogram equalization depends only on the intensity value of the pixel, every pixel with the same gray level is mapped to the same output value. Therefore, histogram equalization is a point operation. So

$$
I_2(x)=M_1(I_1(x))
$$

Since the second operation is a point operation $M_2$,

$$
I_3(x)=M_2(I_2(x))=M_2(M_1(I_1(x)))
$$

Define $M=M_2\circ M_1$, meaning $M(v)=M_2(M_1(v))$. Then

$$
I_3(x)=M(I_1(x))\quad\text{for every pixel }x
$$

Thus, the output at each pixel depends only on its input gray level through the same function $M$, proving that the overall mapping is a point operation.

</div>


## 2. Uniform quantization and qualitative appearance

### <span style="color: green;">Answer</span>


<div style="color: green;">

The mapping $M$ uses the histogram equalization function computed from $I_1$, followed by the quantizer $Q$:

$$
J_3(x)=Q\!\left(H_{I_1}(J_1(x))\right)
$$

The histogram of $I_1$ is concentrated at low gray levels, while $J_1$ contains mostly much brighter pixels. The cumulative histogram of $I_1$ is already close to $1$ at the gray levels appearing in $J_1$. Applying $H_{I_1}$ therefore sends almost all pixels of $J_1$ to values near $255$.

The quantizer has bin width 64, so it maps every intensity in $[192,255]$ to $192$. In particular,

$$
Q(255)=\left\lfloor\frac{255}{64}\right\rfloor\cdot64=192
$$

Thus, $J_3$ will be essentially a uniform light gray image at intensity $192$, and all details will disappearing.

</div>
